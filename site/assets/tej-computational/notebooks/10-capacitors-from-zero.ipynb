## 1. Store a capacitor value
Python uses `e-6` to mean multiply by $10^{-6}$.


```python
# CHANGE the value from 47 to 100
capacitance_uF = 47
capacitance_F = capacitance_uF * 1e-6
assert capacitance_F > 0
print(capacitance_F, "F")
```

## 2. Calculate stored charge
Use $Q=CV$.


```python
voltage_V = 5.0
charge_C = capacitance_F * voltage_V
charge_uC = charge_C * 1e6
print(round(charge_uC, 2), "microcoulombs")
```
