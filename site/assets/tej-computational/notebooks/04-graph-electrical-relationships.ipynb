## Working demonstration


```python
import numpy as np
import matplotlib.pyplot as plt

resistance_ohm = 1000.0
voltage_V = np.linspace(0.0, 12.0, 13)
current_mA = voltage_V / resistance_ohm * 1000.0
assert np.allclose(current_mA, voltage_V)

plt.plot(voltage_V, current_mA, marker='o', label='1.0 kOhm model')
plt.xlabel('Voltage (V)')
plt.ylabel('Current (mA)')
plt.title('Ohmic resistor model')
plt.grid(True)
plt.legend()
plt.show()
```
