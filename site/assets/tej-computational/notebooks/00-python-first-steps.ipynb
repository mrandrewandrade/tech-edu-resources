## Example 4: calculate with variables

Ohm's law is $I=V/R$. Python uses `/` for division. The first answer is in amperes. Multiplying by 1000 converts amperes to milliamperes.


````python
voltage_V = 9.0
resistance_ohm = 1000.0
current_A = voltage_V / resistance_ohm
current_mA = current_A * 1000

print('Current in amperes:', current_A)
print('Current in milliamperes:', current_mA)
````

## Why group values?

Suppose we test a circuit at three voltages. We could make three separate variables:

```python
voltage_1_V = 1.5
voltage_2_V = 3.0
voltage_3_V = 4.5
```

That works, but the values are harder to keep together. A **list** puts the values together under one name.


## Example 5: store several values in a list

Square brackets make a list: `[1.5, 3.0, 4.5]`. Each stored value has a numbered position called an **index**. Python begins at index `0`.

| index | 0 | 1 | 2 |
|---|---:|---:|---:|
| voltage (V) | 1.5 | 3.0 | 4.5 |

Think of the index as the label on a small parts drawer. Index `0` labels the first drawer.


````python
voltage_list_V = [1.5, 3.0, 4.5]
print('Complete list:', voltage_list_V)
print('First value:', voltage_list_V[0])
print('Second value:', voltage_list_V[1])
print('Number of values:', len(voltage_list_V))
````
