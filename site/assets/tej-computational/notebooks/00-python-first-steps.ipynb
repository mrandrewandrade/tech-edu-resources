## Working demonstration 3: give values names

A **variable** is a name that stores a value. The `=` sign stores the value on the right in the name on the left. Read `voltage_V = 9.0` as “store 9.0 in voltage_V.”

## Working demonstration 4: calculate with variables

Ohm's law is $I=V/R$. Python uses `/` for division. The first answer is in amperes. Multiplying by 1000 converts amperes to milliamperes.


```python
voltage_V = 9.0
resistance_ohm = 1000.0
current_A = voltage_V / resistance_ohm
current_mA = current_A * 1000

print('Current in amperes:', current_A)
print('Current in milliamperes:', current_mA)
```

## Working demonstration 5: store several values in a list

A **list** keeps several values together. Square brackets make the list. Python counts positions from zero, so `[0]` selects the first value.


```python
voltage_list_V = [1.5, 3.0, 4.5]
print('Complete list:', voltage_list_V)
print('First value:', voltage_list_V[0])
print('Second value:', voltage_list_V[1])
print('Number of values:', len(voltage_list_V))
```

## Working demonstration 6: meet NumPy

NumPy is a Python library for working with groups of numbers. `import numpy as np` loads the library and gives it the short name `np`. `np.array()` changes a list into a NumPy array.


```python
import numpy as np

voltage_array_V = np.array([1.5, 3.0, 4.5])
print(voltage_array_V)
print(type(voltage_array_V))
```

## One calculation, three answers

NumPy can divide every number in the array by the same resistance. Run this cell.


```python
resistance_ohm = 1000.0
current_array_mA = voltage_array_V / resistance_ohm * 1000
print('Voltage:', voltage_array_V, 'V')
print('Current:', current_array_mA, 'mA')
```
