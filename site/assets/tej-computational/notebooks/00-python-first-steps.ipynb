## Working demonstration 4: calculate with variables

Ohm's law is $I=V/R$. Python uses `/` for division. The first answer is in amperes. Multiplying by 1000 converts amperes to milliamperes.


````python
voltage_V = 9.0
resistance_ohm = 1000.0
current_A = voltage_V / resistance_ohm
current_mA = current_A * 1000

print('Current in amperes:', current_A)
print('Current in milliamperes:', current_mA)
````

## Why group values?

Suppose we test a circuit at three voltages. We could make three separate variables:

```python
voltage_1_V = 1.5
voltage_2_V = 3.0
voltage_3_V = 4.5
```

That works, but the values are harder to keep together. A **container** is one name that holds several values. Python's simplest container for this job is a **list**.


## Working demonstration 5: store several values in a list

Square brackets make a list: `[1.5, 3.0, 4.5]`. Each stored value has a numbered position called an **index**. Python begins at index `0`.

| index | 0 | 1 | 2 |
|---|---:|---:|---:|
| voltage (V) | 1.5 | 3.0 | 4.5 |

Think of the index as the label on a small parts drawer. Index `0` labels the first drawer.


````python
voltage_list_V = [1.5, 3.0, 4.5]
print('Complete list:', voltage_list_V)
print('First value:', voltage_list_V[0])
print('Second value:', voltage_list_V[1])
print('Number of values:', len(voltage_list_V))
````

## From a list to an array

A **NumPy array** is another container. In this lesson it is a numbered row of values, like the list above. The useful difference is that NumPy is designed to perform the same calculation on every number in the container.

| Python list | NumPy array |
|---|---|
| groups values together | groups values together |
| can hold different kinds of values | normally holds one kind of value |
| good for general storage | good for repeated numerical calculations |

For now, an array does **not** mean a matrix or a complicated table. It is just one ordered row of three voltage numbers.


## Working demonstration 6: build the first NumPy array

`import numpy as np` loads NumPy and gives it the short name `np`. `np.array()` copies the three list values into a NumPy array.


````python
import numpy as np

voltage_list_V = [1.5, 3.0, 4.5]
voltage_array_V = np.array(voltage_list_V)
print('List:', voltage_list_V)
print('Array:', voltage_array_V)
print('First array value:', voltage_array_V[0])
print(type(voltage_array_V))
````

### Predict before running

````python
doubled_voltage_V = voltage_array_V * 2
print('Original array:', voltage_array_V)
print('Doubled array:', doubled_voltage_V)
````

## One calculation, three answers

NumPy divides **each** voltage by the same resistance. The first voltage produces the first current, the second voltage produces the second current, and the third voltage produces the third current.


````python
resistance_ohm = 1000.0
current_array_mA = voltage_array_V / resistance_ohm * 1000
print('Voltage:', voltage_array_V, 'V')
print('Current:', current_array_mA, 'mA')
````
