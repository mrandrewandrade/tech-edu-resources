## Working demonstration
Use Python as a calculator and a check.


```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Calculate tau


```python
R = 4_700
C = 220e-6
tau = R * C
assert tau > 0
print(round(tau, 3), "s")
```

## 3. Test three capacitor choices


```python
R = 10_000
choices_uF = np.array([100, 220, 330])
choices_F = choices_uF * 1e-6
taus = R * choices_F
print(choices_uF)
print(taus)
```

## 4. Draw all three charging curves


```python
t = np.linspace(0, 18, 300)
for C_uF, this_tau in zip(choices_uF, taus):
    vc = 5 * (1 - np.exp(-t/this_tau))
    plt.plot(t, vc, label=f"{C_uF} uF")
plt.xlabel("time (s)")
plt.ylabel("capacitor voltage (V)")
plt.grid()
plt.legend()
plt.show()
```
