## 6. Solve and calculate residuals

[`numpy.linalg.solve`](https://numpy.org/doc/stable/reference/generated/numpy.linalg.solve.html) solves $A\mathbf{x}=\mathbf{b}$. The residual vector is $A\mathbf{x}-\mathbf{b}$.


```python
currents_mA = None
residual = None

if A is None or b is None:
    print('Complete A and b first.')
else:
    currents_mA = np.linalg.solve(A, b)
    residual = A @ currents_mA - b
    for name, value in zip(['I1', 'I2', 'I3'], currents_mA):
        print(f'{name} = {value:.6f} mA')
    print('Residuals:', residual)
    assert np.allclose(residual, 0.0, atol=1e-10)
```
