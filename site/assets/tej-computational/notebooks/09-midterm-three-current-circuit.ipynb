## 7. Enter and solve the matrix

Replace both `None` values. Keep row order Loop 1, Loop 2 and variable order $[I_1,I_2]$.

Example syntax: `np.array([[1.0, 2.0], [3.0, 4.0]])`


````python
A = None  # Replace with the 2 by 2 coefficient array.
b = None  # Replace with the two right-side values.

if A is None or b is None:
    currents_mA = None
    print('Complete A and b, then run this cell again.')
else:
    assert A.shape == (2, 2), 'A must have 2 rows and 2 columns.'
    assert b.shape == (2,), 'b must contain 2 values.'
    currents_mA = np.linalg.solve(A, b)
    I1_mA, I2_mA = currents_mA
    I3_mA = I1_mA - I2_mA
    residual_V = A @ currents_mA - b
    print(f'I1 = {I1_mA:.3f} mA')
    print(f'I2 = {I2_mA:.3f} mA')
    print(f'I3 = {I3_mA:.3f} mA')
    print('Residuals:', residual_V)
    assert np.allclose(residual_V, 0.0, atol=1e-10)
    assert np.allclose([I1_mA, I2_mA, I3_mA], [2.0, 1.0, 1.0])
    print('CHECKED: the matrix produces the Life Checkpoint currents.')
````

## 8. Make the current dashboard

The solved currents become a graph. Change the graph title and colours if you want. Do not change the current values.


````python
dashboard_title = f"{display_name}'s Life Checkpoint"
bar_colours = ['#6a00ff', '#00b7ff', '#ffcc00']

if currents_mA is None:
    print('Solve the matrix first.')
else:
    dashboard_currents = np.array([I1_mA, I2_mA, I3_mA])
    fig, ax = plt.subplots(figsize=(7, 4))
    bars = ax.bar(['I1', 'I2', 'I3'], dashboard_currents, color=bar_colours, edgecolor='black')
    ax.bar_label(bars, fmt='%.1f mA', padding=4, fontsize=12)
    ax.set_ylim(0, max(dashboard_currents) + 1.0)
    ax.set_ylabel('Current (mA)')
    ax.set_title(dashboard_title, weight='bold')
    ax.grid(axis='y', alpha=0.25)
    plt.show()
````
