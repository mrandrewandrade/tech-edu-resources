## Working demonstration

Inspect how a 5% resistor tolerance changes the possible current range.


```python
# Working demonstration: tolerance range for a nominal resistor.
import numpy as np
import matplotlib.pyplot as plt

source_V = 5.0
nominal_ohm = 1000.0
tolerance = 0.05
resistance_ohm = np.linspace(nominal_ohm * (1-tolerance), nominal_ohm * (1+tolerance), 101)
current_mA = source_V / resistance_ohm * 1000.0
assert current_mA.min() > 0
print(f'Current range: {current_mA.min():.3f} to {current_mA.max():.3f} mA')
plt.plot(resistance_ohm, current_mA)
plt.xlabel('Resistance (ohm)')
plt.ylabel('Current (mA)')
plt.title('Effect of 5% resistor tolerance')
plt.grid(True)
plt.show()
```
