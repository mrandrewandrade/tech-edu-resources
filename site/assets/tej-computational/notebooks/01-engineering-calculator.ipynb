## Working demonstration 1: variables, arithmetic, and output

Run this cell unchanged first.


```python
voltage_V = 9.0
resistance_ohm = 1000.0
current_A = voltage_V / resistance_ohm
power_W = voltage_V * current_A

current_mA = current_A * 1000
power_mW = power_W * 1000

print(f'Current: {current_mA:.2f} mA')
print(f'Power: {power_mW:.1f} mW')
assert abs(current_A - 0.009) < 1e-12
assert abs(power_W - 0.081) < 1e-12
```

## Working demonstration 2: lists, indexing, and a loop

A list stores several related values. Python starts counting positions at zero.


```python
resistor_values_ohm = [1000.0, 2200.0, 3300.0, 4700.0]
print('First value:', resistor_values_ohm[0])
print('Number of values:', len(resistor_values_ohm))

for value_ohm in resistor_values_ohm:
    current_mA = 9.0 / value_ohm * 1000
    print(f'{value_ohm:6.0f} ohm -> {current_mA:5.2f} mA')
```

## Working demonstration 3: NumPy arrays

NumPy performs the same operation on a complete array without writing a separate calculation for every value. Inspect `shape`, `dtype`, and the first and last results before trusting the output.


```python
import numpy as np

voltage_V = np.array([1.5, 3.0, 4.5, 6.0, 7.5, 9.0])
resistance_ohm = 2200.0
current_mA = voltage_V / resistance_ohm * 1000
power_mW = voltage_V * current_mA

print('shape:', voltage_V.shape)
print('dtype:', voltage_V.dtype)
print('current (mA):', current_mA)
print('minimum and maximum:', current_mA.min(), current_mA.max())
print('mean:', current_mA.mean())

assert voltage_V.shape == current_mA.shape
assert np.all(current_mA > 0)
assert np.all(np.diff(current_mA) > 0)
```

## Working demonstration 4: write and test a function


```python
def calculate_current_mA(voltage_V, resistance_ohm):
    """Return resistor current in milliamperes."""
    return np.asarray(voltage_V) / resistance_ohm * 1000

test_current_mA = calculate_current_mA([3.0, 6.0, 9.0], 3000.0)
print(test_current_mA)
assert np.allclose(test_current_mA, [1.0, 2.0, 3.0])
```
