## Working demonstration

Run the cell unchanged first. Then explain each unit conversion.


```python
voltage_V = 9.0
resistance_ohm = 1000.0
current_A = voltage_V / resistance_ohm
power_W = voltage_V * current_A

print(f'Current: {current_A * 1000:.2f} mA')
print(f'Power: {power_W * 1000:.1f} mW')
assert abs(current_A - 0.009) < 1e-12
assert abs(power_W - 0.081) < 1e-12
```
