## Working demonstration: from KVL to a matrix

Use the completed Question 4 equations:

$$5I_1-2I_2=10$$

$$-2I_1+7I_2=6.$$

Keep the variable order $[I_1,I_2]$ in every row. Then

$$
\underbrace{\begin{bmatrix}5&-2\\-2&7\end{bmatrix}}_{A}\underbrace{\begin{bmatrix}I_1\\I_2\end{bmatrix}}_{\mathbf{x}}=\underbrace{\begin{bmatrix}10\\6\end{bmatrix}}_{\mathbf{b}}.
$$

With resistance in $\mathrm{k\Omega}$ and voltage in $\mathrm{V}$, the currents are in $\mathrm{mA}$.


```python
import numpy as np
try:
    import matplotlib.pyplot as plt
except ImportError:
    plt = None
    print('Matplotlib is unavailable; use the numerical table for the investigation.')

variable_names = np.array(['I1', 'I2'])
A_demo = np.array([[5.0, -2.0], [-2.0, 7.0]])  # kOhm
b_demo = np.array([10.0, 6.0])                  # V
```

```python
currents_demo_mA = np.linalg.solve(A_demo, b_demo)
residual_demo_V = A_demo @ currents_demo_mA - b_demo
for name, value in zip(variable_names, currents_demo_mA):
    print(f'{name} = {value:.6f} mA')
print('Residuals (V):', residual_demo_V)
print('Maximum absolute residual (V):', np.max(np.abs(residual_demo_V)))
assert np.allclose(residual_demo_V, 0.0, atol=1e-10)
```

### Read the output

- A positive current agrees with the chosen arrow.
- A negative current means the real current is opposite the chosen arrow.
- Each residual is the left side minus the right side of one entered equation.


## Documentation-driven debugging

Engineers are not expected to memorize every function. They are expected to reproduce a problem, read the exact message, inspect the inputs, find the relevant documentation, and verify the repair.

Use this protocol before asking for the corrected line of code:

1. **Reproduce:** run the smallest cell that still shows the problem.
2. **Read:** copy the final exception line or describe the suspicious output precisely.
3. **Inspect:** print `type(...)`, `.shape`, units, and the values entering the function.
4. **Research:** try `help(function)`, `function?`, or search the exact error with the library and function name. Prefer the official NumPy or Jupyter documentation.
5. **Repair:** change the smallest necessary part and explain why the documentation supports it.
6. **Verify:** rerun from a clean kernel and use an independent equation, unit check, or residual.

Your debugging record must include the exact error or wrong result, search terms, source consulted, the requirement you found, the repair, and the verification evidence.

## Parameter investigation: predict before running

Predict what will happen to $I_1$ and $I_2$ as the left-loop source rises from $6\,\mathrm{V}$ to $14\,\mathrm{V}$.

**Prediction:** ___


```python
# STEP 3 — Change one source systematically and graph the response.
source_row = 0
source_values_V = np.linspace(6.0, 14.0, 9)
sweep_currents_mA = []
for source_V in source_values_V:
    b_changed = student_b.copy()
    b_changed[source_row] = source_V
    sweep_currents_mA.append(np.linalg.solve(student_A, b_changed))
sweep_currents_mA = np.array(sweep_currents_mA)

print('Source (V)    I1 (mA)    I2 (mA)')
for source_V, row in zip(source_values_V, sweep_currents_mA):
    print(f'{source_V:8.2f} {row[0]:10.4f} {row[1]:10.4f}')

if plt is not None:
    plt.figure(figsize=(8, 5))
    for column, name in enumerate(student_variable_names):
        plt.plot(source_values_V, sweep_currents_mA[:, column], marker='o', label=name)
    plt.axhline(0.0, color='black', linewidth=0.8)
    plt.xlabel('Changed source voltage (V)')
    plt.ylabel('Solved current (mA)')
    plt.title('How the circuit solution changes with the source')
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()
```
