## Working demonstration

Upload `circuit-measurements.csv` beside this notebook. Preserve the raw file unchanged.


```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

csv_path = Path('circuit-measurements.csv')
if not csv_path.exists():
    csv_path = Path('../data/circuit-measurements.csv')
data = pd.read_csv(csv_path)
required = {'voltage_V', 'resistance_ohm', 'measured_current_mA'}
assert required.issubset(data.columns)
assert not data[list(required)].isna().any().any()
data['predicted_current_mA'] = data['voltage_V'] / data['resistance_ohm'] * 1000.0
data['absolute_difference_mA'] = (data['measured_current_mA'] - data['predicted_current_mA']).abs()
reference = data['predicted_current_mA'].abs()
data['percentage_difference'] = np.where(reference > 0, data['absolute_difference_mA'] / reference * 100.0, np.nan)
display(data)
```

```python
plt.scatter(data['voltage_V'], data['measured_current_mA'], label='Measured')
plt.plot(data['voltage_V'], data['predicted_current_mA'], label='Predicted')
plt.xlabel('Voltage (V)')
plt.ylabel('Current (mA)')
plt.title('Measured and predicted current')
plt.grid(True)
plt.legend()
plt.show()
assert data['absolute_difference_mA'].max() < 0.2
```
