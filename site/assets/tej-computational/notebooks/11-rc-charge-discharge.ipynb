## Working demonstration
The NumPy and exponential code is already written. You are not expected to write it.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Set the circuit values
Keep ohms with farads.


```python
R = 10_000
C = 100e-6
Vs = 5.0
tau = R * C
assert tau > 0
print("time constant =", tau, "s")
```

## 2. Run the supplied model
`np.linspace` makes the time values and `np.exp` makes the smooth curve. Run this cell without changing it.


```python
t = np.linspace(0, 5*tau, 200)
charge = Vs * (1 - np.exp(-t/tau))
discharge = Vs * np.exp(-t/tau)
```

## 3. Plot both curves


```python
plt.plot(t, charge, label="charging")
plt.plot(t, discharge, label="discharging")
plt.xlabel("time (s)")
plt.ylabel("capacitor voltage (V)")
plt.grid()
plt.legend()
plt.show()
```
