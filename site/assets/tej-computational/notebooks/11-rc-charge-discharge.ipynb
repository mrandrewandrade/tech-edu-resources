## 2. Store Tuesday's measurements
Each array contains one column from Tuesday's table.

```python
time_s = np.array([0, 5, 10, 15, 20, 25])
charging_v = np.array([0.0, 2.0, 3.2, 4.0, 4.5, 4.8])
discharging_v = np.array([5.0, 3.0, 1.8, 1.0, 0.5, 0.2])

assert len(time_s) == len(charging_v) == len(discharging_v)
print("time:", time_s)
print("charging:", charging_v)
print("discharging:", discharging_v)
```

## 3. Calculate the time constant
For an RC circuit,

$$\tau=RC$$

Use ohms for $R$ and farads for $C$. The result is in seconds. Here, $R=10\,000\,\Omega$ and $C=1000\,\mu\text{F}=0.001\,\text{F}$.

```python
R = 10_000       # ohms
C = 1000e-6      # farads
Vs = 5.0         # volts
tau = R * C

assert tau > 0
print("tau =", tau, "s")
```

## 4. Working demonstration: define two functions
A Python function has a name, inputs in brackets, and a returned output. The input `t` can be one time or an entire NumPy array.

Charging: $V_C(t)=V_S(1-e^{-t/\tau})$

Discharging: $V_C(t)=V_0e^{-t/\tau}$

```python
def charging_voltage(t, supply_v, time_constant):
    return supply_v * (1 - np.exp(-t / time_constant))

def discharging_voltage(t, start_v, time_constant):
    return start_v * np.exp(-t / time_constant)
```

## 7. Draw the function curves over the dots
`np.linspace` makes 200 time inputs. The functions return 200 predicted voltages. Run the cell and compare the smooth curves with the measured dots.

**Common errors:** check the function name, the order of its three inputs, and every comma and bracket.

```python
smooth_time = np.linspace(0, 25, 200)
charging_curve = charging_voltage(smooth_time, Vs, tau)
discharging_curve = discharging_voltage(smooth_time, Vs, tau)

plt.scatter(time_s, charging_v, color="blue", marker="o", label="charging data")
plt.scatter(time_s, discharging_v, color="red", marker="s", label="discharging data")
plt.plot(smooth_time, charging_curve, color="blue", label="charging function")
plt.plot(smooth_time, discharging_curve, color="red", label="discharging function")
plt.xlabel("time (s)")
plt.ylabel("capacitor voltage (V)")
plt.title("RC data and function model")
plt.ylim(0, 5.5)
plt.grid()
plt.legend()
plt.show()
```

## 8. Predict the fading LED
The first physical circuit will let the charged capacitor power a red LED through a resistor. During discharge, KVL gives

$$V_C=V_R+V_{LED}$$

Using a simple fixed forward-voltage model,

$$I_{LED}=\frac{V_C-V_{LED}}{R_{LED}}$$

The function below prevents the model from reporting negative current after the capacitor voltage falls below the LED forward voltage. This is a first model. The physical circuit also uses a 10 kOhm bleeder resistor so the capacitor finishes discharging safely.

```python
led_forward_v = 2.0
led_resistance = 1000

def led_current_mA(capacitor_v, forward_v, resistance):
    current_A = (capacitor_v - forward_v) / resistance
    return np.maximum(current_A, 0) * 1000

led_current_curve = led_current_mA(discharging_curve, led_forward_v, led_resistance)

plt.plot(smooth_time, led_current_curve, color="purple")
plt.xlabel("time after power is opened (s)")
plt.ylabel("modelled LED current (mA)")
plt.title("Why the LED fades as the capacitor discharges")
plt.ylim(bottom=0)
plt.grid()
plt.show()
```
