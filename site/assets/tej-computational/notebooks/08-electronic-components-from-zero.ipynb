## Model boundary

This notebook is a start-now laboratory when physical parts are unavailable. A model can support a prediction, comparison, or design choice. It cannot identify a real pinout, inspect soldering, verify a rating, or prove that a physical circuit is safe.


## Working demonstration 1: resistor current and power


```python
import numpy as np
import matplotlib.pyplot as plt

source_V = 5.0
resistance_ohm = np.array([220.0, 330.0, 470.0, 680.0, 1000.0])
current_mA = source_V / resistance_ohm * 1000.0
power_mW = source_V**2 / resistance_ohm * 1000.0
assert np.all(np.diff(current_mA) < 0)
for R, I, P in zip(resistance_ohm, current_mA, power_mW):
    print(f'{R:5.0f} ohm: {I:6.2f} mA, {P:6.2f} mW')
```

A larger resistance produces less current at the same voltage. This calculation does not automatically approve a resistor's tolerance, power rating, temperature, or package.


## Working demonstration 2: capacitor charge and energy


```python
capacitance_F = np.array([33e-12, 100e-9, 10e-6, 100e-6])
capacitor_voltage_V = 5.0
charge_C = capacitance_F * capacitor_voltage_V
energy_J = 0.5 * capacitance_F * capacitor_voltage_V**2
assert np.all(charge_C > 0) and np.all(energy_J > 0)
for C, Q, E in zip(capacitance_F, charge_C, energy_J):
    print(f'{C:.3e} F: Q={Q:.3e} C, E={E:.3e} J')
```

The 33 pF value matches the scale of the XL741 compensation capacitor. It stores much less energy than a 100 µF capacitor at the same voltage, but it can still have an important signal-stability role.


## Working demonstration 3: LED resistor choices


```python
led_source_V = 5.0
led_forward_V_model = 2.0
candidate_ohm = np.array([220.0, 330.0, 470.0, 680.0, 1000.0])
led_current_mA = (led_source_V - led_forward_V_model) / candidate_ohm * 1000.0
target_mA = 10.0
best_index = np.argmin(np.abs(led_current_mA - target_mA))
print(f'Modelled closest choice: {candidate_ohm[best_index]:.0f} ohm at {led_current_mA[best_index]:.2f} mA')
assert np.all(led_current_mA > 0)

plt.plot(candidate_ohm, led_current_mA, marker='o')
plt.axhline(target_mA, color='tab:orange', linestyle='--', label='target')
plt.xlabel('Series resistance (ohm)')
plt.ylabel('Modelled LED current (mA)')
plt.title('LED current using a fixed forward-voltage model')
plt.grid(True)
plt.legend()
plt.show()
```

The fixed forward-voltage model is only a first estimate. A real LED's forward voltage depends on device, current, temperature, and manufacturing variation.


## Working demonstration 4: ideal transistor switch


```python
input_state = np.array([0, 1])  # 0=off, 1=on in this deliberately simple model
load_resistance_ohm = 470.0
load_supply_V = 5.0
ideal_load_current_mA = input_state * load_supply_V / load_resistance_ohm * 1000.0
assert ideal_load_current_mA[0] == 0
assert ideal_load_current_mA[1] > 0
print('Ideal switch load currents (mA):', ideal_load_current_mA)
```

This two-state model omits base or gate drive, saturation voltage, leakage, gain variation, switching time, thermal limits, and pinout. Use it to understand the control idea, not to approve hardware.


## Working demonstration 5: op-amp follower and modelled saturation


```python
follower_input_V = np.linspace(-15.0, 15.0, 301)
assumed_output_limit_V = 12.0  # explicit model assumption, not an XL741 specification
ideal_follower_V = follower_input_V.copy()
limited_follower_V = np.clip(follower_input_V, -assumed_output_limit_V, assumed_output_limit_V)
assert np.all(np.abs(limited_follower_V) <= assumed_output_limit_V)

plt.plot(follower_input_V, ideal_follower_V, '--', label='ideal Vout = Vin')
plt.plot(follower_input_V, limited_follower_V, label='example limited-output model')
plt.xlabel('Input voltage (V)')
plt.ylabel('Output voltage (V)')
plt.title('Ideal follower and an explicit saturation assumption')
plt.grid(True)
plt.legend()
plt.show()
```

The limit above is chosen only to show saturation. Replace it with verified datasheet or measured values for a specific device and operating condition.
