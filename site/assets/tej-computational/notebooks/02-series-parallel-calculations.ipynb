## Working demonstration: series circuit


```python
source_V = 15.0
series_ohm = [1000.0, 2200.0, 3300.0]
total_ohm = sum(series_ohm)
total_A = source_V / total_ohm
drops_V = [total_A * value for value in series_ohm]
print(total_ohm, total_A * 1000, drops_V)
assert abs(sum(drops_V) - source_V) < 1e-12
```

## Working demonstration: parallel circuit


```python
parallel_ohm = [1000.0, 2200.0, 3300.0]
equivalent_ohm = 1 / sum(1 / value for value in parallel_ohm)
branch_A = [source_V / value for value in parallel_ohm]
total_parallel_A = source_V / equivalent_ohm
assert abs(sum(branch_A) - total_parallel_A) < 1e-12
assert equivalent_ohm < min(parallel_ohm)
print(f'Equivalent resistance: {equivalent_ohm:.2f} ohm')
```
